# Thực nghiệm Regression

Notebook khung, chưa có kết quả thực nghiệm.

1. Đọc Train và `models/preprocessing_setup.joblib`.
2. Thống nhất bốn thuật toán Regression; ghép từng mô hình với bộ tiền xử lý phù hợp bằng Pipeline.
3. So sánh bằng cùng 5-fold KFold, shuffle=True, random_state=42. Dùng RMSE làm tiêu chí chính; báo cáo thêm MAE, R² và độ lệch chuẩn CV.
4. Tuning các mô hình triển vọng bằng CV trên Train. Chưa sử dụng Test.
5. Lưu cấu hình mô hình được chọn vào `models/regression_best_config.json`.

Kết quả dự kiến: `results/regression/metrics/` và `results/regression/figures/`. Không tạo kết quả giả hay mặc định thuật toán thắng trước khi chạy.
