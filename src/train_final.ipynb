# Huấn luyện cuối và đánh giá Test

Notebook khung, chưa huấn luyện mô hình.

1. Đọc cấu hình Regression đã chọn bằng CV và cấu hình tiền xử lý.
2. Fit Pipeline trên toàn bộ Train; không tuning lại dựa vào Test.
3. Dự đoán Test và báo cáo MAE, RMSE, R²; vẽ Actual vs Predicted và phân tích phần dư.
4. Dùng cùng ngưỡng `cost_thresholds` đã lưu để phân mức chi phí thực tế và dự đoán. Khoảng Thấp gồm ngưỡng dưới, Trung bình gồm ngưỡng trên, Cao lớn hơn ngưỡng trên.
5. Đánh giá phân mức bằng confusion matrix và F1-macro; phân tích các trường hợp gần ngưỡng. Đây là kết quả suy ra từ Regression, không phải classifier riêng.
6. Lưu `models/regression_pipeline.joblib` và `models/cost_thresholds.json`. Tệp JSON chứa ngưỡng, nhãn, nguồn Train và quy ước biên để web áp dụng nhất quán.

Metrics và biểu đồ được lưu trong `results/regression/metrics/` và `results/regression/figures/`.
