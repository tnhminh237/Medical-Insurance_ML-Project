# Chia Train/Test
Chia dữ liệu sạch theo tỷ lệ 80/20, cố định `random_state=42` để cả nhóm dùng chung.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

ROOT = Path.cwd().resolve()
if ROOT.name == "src":
    ROOT = ROOT.parent
DATA_DIR = ROOT / "data"

data = pd.read_csv(DATA_DIR / "medical_insurance_cleaned.csv")
```

## 1. Chia dữ liệu

```python
train, test = train_test_split(data, test_size=0.2, random_state=42)

print(f"Train: {len(train):,} dòng")
print(f"Test: {len(test):,} dòng")
```

## 2. Lưu hai tập
Dùng Train cho EDA và lựa chọn mô hình; giữ Test cho đánh giá cuối cùng.

```python
train.to_csv(DATA_DIR / "medical_insurance_train.csv", index=False)
test.to_csv(DATA_DIR / "medical_insurance_test.csv", index=False)
print("Đã lưu medical_insurance_train.csv và medical_insurance_test.csv")
```
