# XAI và phân tích sai số Regression

Notebook khung, chưa có kết quả XAI.

- Tải Pipeline hồi quy cuối cùng, giải thích bằng SHAP hoặc Feature Importance phù hợp thuật toán.
- Phân biệt tên đặc trưng gốc với các cột sau encoding khi diễn giải.
- Giải thích toàn cục và một số dự đoán cụ thể.
- Phân tích sai số theo mức chi phí và nhóm đặc trưng; chú ý trường hợp chi phí cao và gần ngưỡng phân mức.
- Không sử dụng nhận xét từ Test để chọn lại mô hình rồi báo cáo trên chính Test đó.

Lưu đầu ra trong `results/xai/figures/` và `results/xai/metrics/` khi có kết quả thực tế.
